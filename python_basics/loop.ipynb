### **Loops (반복문)**
- 특정 코드를 정해진 횟수나 특정 조건에 따라 반복적으로 실행할 때 사용합니다.
- 파이썬에서는 주로 `for`문과 `while`문을 사용합니다.

### **1. for 반복문**
- 리스트, 튜플, 문자열 등 순회 가능한(iterable) 객체의 요소를 하나씩 꺼내어 코드를 실행합니다.
- **기본 구조**
```python
for 변수 in 순회가능한객체:
    반복할 코드
```

````python
fruits = ['사과', '바나나', '딸기']

for fruit in fruits:
    print(f'나는 {fruit}를 좋아해!')
````

#### **`range()` 함수와 함께 사용하기**
- `range()` 함수는 특정 범위의 숫자 시퀀스를 생성해주어 `for`문과 자주 함께 쓰입니다.
- `range(끝)`: 0부터 `끝-1`까지의 숫자를 생성합니다.
- `range(시작, 끝)`: `시작`부터 `끝-1`까지의 숫자를 생성합니다.
- `range(시작, 끝, 단계)`: `시작`부터 `끝-1`까지 `단계`만큼 건너뛰며 숫자를 생성합니다.

````python
# 0부터 4까지 출력
for i in range(5):
    print(i, end=' ')

print('\n' + '-'*20)

# 1부터 5까지 출력
for i in range(1, 6):
    print(i, end=' ')

print('\n' + '-'*20)

# 2부터 10까지 2씩 건너뛰며 출력 (짝수)
for i in range(2, 11, 2):
    print(i, end=' ')
````

#### **중첩 for문 (Nested for loops)**
- 반복문 안에 또 다른 반복문을 사용하는 구조입니다.
- 대표적인 예로 구구단 출력이 있습니다.

````python
# 구구단 출력하기
for i in range(2, 10):
    print(f'--- {i}단 ---')
    for j in range(1, 10):
        print(f'{i} X {j} = {i*j}')
    print()
````

### **2. while 반복문**
- 주어진 `조건`이 참(True)인 동안 코드를 반복적으로 실행합니다.
- 조건이 거짓(False)이 되면 반복을 멈춥니다.
- **기본 구조**
```python
while 조건:
    반복할 코드
```
- **주의**: 조건이 계속 참으로 유지되면 **무한 루프**에 빠질 수 있으므로, 루프 내에서 조건을 변경하는 코드가 반드시 필요합니다.

````python
count = 5

while count > 0:
    print(f'카운트다운: {count}')
    count -= 1 # 이 코드가 없으면 무한 루프에 빠짐

print('발사!')
````

### **3. 반복문 제어**
- 반복문의 흐름을 제어해야 할 때가 있습니다.
- `break`: 반복문을 즉시 완전히 중단합니다.
- `continue`: 현재 반복을 건너뛰고 다음 반복을 시작합니다.

#### **`break` 예제**
- 리스트에서 특정 값을 찾으면 반복을 멈춥니다.

````python
numbers = [1, 5, 2, -1, 9, 4]

for num in numbers:
    if num < 0:
        print('음수를 발견하여 반복을 중단합니다.')
        break
    print(f'현재 숫자: {num}')
````

#### **`continue` 예제**
- 1부터 10까지의 숫자 중 홀수만 출력합니다.

````python
for i in range(1, 11):
    if i % 2 == 0: # 숫자가 짝수이면
        continue     # 아래 print를 실행하지 않고 다음 반복으로 넘어감
    print(i)
````
