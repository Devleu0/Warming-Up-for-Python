## Ⅰ. 기본 내장 함수 (Built-in Functions)

모듈 임포트(`import`) 없이 바로 사용할 수 있는 파이썬의 기본 함수들입니다.

### `print()`
값을 화면에 출력합니다.

```python
print("Hello, Jupyter!")
```

### `input()`
사용자로부터 문자열을 입력받습니다. (노트북 환경에서는 입력창이 나타납니다)

### `len()`
객체(문자열, 리스트 등)의 길이를 반환합니다.

```python
my_list = [1, 2, 3, 4, 5]
print(f"리스트의 길이: {len(my_list)}")

my_string = "Python"
print(f"문자열의 길이: {len(my_string)}")
```

### `type()`
객체의 자료형을 확인합니다.

```python
print(type(123))          # 정수형
print(type(3.14))         # 실수형
print(type("Hello"))      # 문자열
print(type([1, 2, 3]))    # 리스트
```

### `int()`, `float()`, `str()`, `list()`
값을 다른 자료형으로 변환합니다.

```python
num_str = "100"
num_int = int(num_str)
print(f"문자열 '{num_str}' -> 정수 {num_int}")

pi_float = 3.14
pi_str = str(pi_float)
print(f"실수 {pi_float} -> 문자열 '{pi_str}'")

text = "abc"
char_list = list(text)
print(f"문자열 '{text}' -> 리스트 {char_list}")
```

### `sum()`, `max()`, `min()`
숫자로 이루어진 반복 가능한 객체(리스트 등)의 합계, 최댓값, 최솟값을 구합니다.

```python
numbers = [10, 5, 8, 20, 3]

print(f"합계: {sum(numbers)}")
print(f"최댓값: {max(numbers)}")
print(f"최솟값: {min(numbers)}")
```

### `sorted()`
객체를 정렬한 **새로운** 리스트를 반환합니다. (원본은 변경되지 않습니다)

```python
original_list = [3, 1, 4, 1, 5, 9, 2]
sorted_list = sorted(original_list)

print(f"원본 리스트: {original_list}")
print(f"정렬된 새 리스트: {sorted_list}")
```

## Ⅱ. 자료형별 내장 함수 (메서드)
각 자료형이 가지고 있는 고유한 기능의 함수(메서드)들입니다.

### 1. 문자열 (String)

```python
s = "  Hello, Python World!  "

# .upper() / .lower()
print(f"대문자: {s.upper()}")
print(f"소문자: {s.lower()}")

# .strip()
print(f"공백 제거: '{s.strip()}'")

# .replace()
print(f"문자열 교체: {s.replace('Python', 'Jupyter')}")

# .split()
words = "apple,banana,cherry"
print(f"문자열 나누기: {words.split(',')}")

# .join()
char_list = ['p', 'y', 't', 'h', 'o', 'n']
print(f"문자열 합치기: {'-'.join(char_list)}")
```

### 2. 리스트 (List)

```python
fruits = ['apple', 'banana']
print(f"초기 리스트: {fruits}")

# .append()
fruits.append('cherry')
print(f"append 후: {fruits}")

# .pop()
popped_item = fruits.pop()
print(f"pop으로 제거된 항목: {popped_item}")
print(f"pop 후: {fruits}")

# .sort() (원본 변경)
numbers = [3, 1, 4, 2]
numbers.sort()
print(f"sort 후: {numbers}")

# .reverse() (원본 변경)
numbers.reverse()
print(f"reverse 후: {numbers}")

# .insert()
numbers.insert(1, 99) # 1번 인덱스에 99 삽입
print(f"insert 후: {numbers}")
```

### 3. 딕셔너리 (Dictionary)

```python
person = {
    'name': '홍길동',
    'age': 30,
    'city': '서울'
}

# .keys()
print(f"키: {person.keys()}")

# .values()
print(f"값: {person.values()}")

# .items() (for 문과 함께 자주 사용)
print(f"항목들: {person.items()}")
for key, value in person.items():
    print(f"  - {key}: {value}")

# .get()
print(f"이름: {person.get('name')}")
print(f"직업: {person.get('job', '알 수 없음')} # 키가 없을 때 기본값 반환")
```

## Ⅲ. 파이썬 자료구조: 리스트, 딕셔너리, 튜플, 셋

### **리스트(list) & 딕트(dict)**
 * 복수개의 값을 담을 수 있는 데이터 구조
 * 실생활에서 사용하는 리스트(학생 리스트, 성적 리스트 등등)과 동일한 의미로 이해
 * list, dict, set - mutable    (생성된 후에 변경 가능)  
 * tuple - immutable (생성된 후에 변경 불가능)  

#### **리스트 초기화**
 - [] 안에 값을 담아서 생성
 - list() 함수로 생성
 - str.split()함수로 생성

```python
a = []
print(a)

a = [1, 2, 3, 5, 10]
print(a)
```

```python
a = list()
print(a)
```

```python
a = 'hello world'.split()
print(a)
```

* **list() 함수**
 + 다른 데이터 타입을 리스트로 변환할 때도 사용

```python
a = 'hello world'
b = list(a)
print(b)
```

+ **string split 함수**
 * 구분자로 구분되는 리스트를 반환 (빈번히 사용 됨)

```python
a = 'hello,world,python'
b = a.split(',')
print(b)
```

#### **리스트 indexing**
 * 문자열의 인덱싱과 동일하게 동작
 * [] 연산자를 이용하여 항목 얻어오기
 * [i] - i번째 원소를 반환
 * i가 음수인 경우도 가능하며 마지막원소가 -1로 하여 앞으로 갈때마다 1씩 감소함

```python
a = [1, 2, 3, 4, 5, 6]
print(a[0])
print(a[3])
print(a[-1])
```

#### **리스트 개별 아이템에 접근**
 - 인덱스에 접근하여 값을 업데이트 가능

```python
a = [1, 2, 3, 4, 5]
a[0] = 100
a[-1] = 99
print(a)
```

#### **리스트 slicing**
 - 문자열 슬라이싱과 동일하게 동작
 - 슬라이싱의 결과 역시 list!

```python
a = [0, 1, 2, 3, 4, 5, 6, 7, 8]
print(a[0:4])
print(a[:5])
print(a[3:])
print(a[0:7:2])
```

### list 멤버 함수
 - 생성된 리스트 객체에 동작하는 함수
 - 향후, 클래스와 멤버 함수 개념을 이해할 예정

* **append()**
 + 리스트의 끝에 항목을 추가함

```python
a = [1, 2, 3, 4]
a.append(10)
print(a)
```

+ **extend()**
 * 리스트를 연장
 * += 로도 가능함

```python
a = [1, 2, 3]
b = [3, 4, 5]
a.extend(b)
print(a)

c = [1, 2, 3]
c += [3, 4, 5]
print(c)
```

+ **insert()로 항목추가**
 * 리스트의 원하는 위치에 추가 가능
 * 앞에 인덱스를, 뒤에 아이템을 명시

```python
a = [1, 3, 4, 5]
a.insert(1, 2)
print(a)
```

+ **remove()**
 * 값으로 항목 삭제(인덱스 아님)
 * 지우려는 값이 2개 있으면 가장 앞에 값만 삭제
 * 내가 지우려는 값이 list에 없으면 오류 발생

```python
a = [1, 2, 3, 4, 2]
a.remove(2)
print(a)
```

+ **pop()**
  - 지우고자 하는 아이템을 반환 후, 삭제
  - 인덱스로 삭제
  - 인덱스를 명시하지 않으면 가장 마지막 값이 삭제 됨

```python
a = [1, 2, 3, 4, 5]
d = a.pop(2)
print(a)
print(d)

e = a.pop()
print(a)
print(e)
```

+ **index()**
 * 찾고자 하는 값의 인덱스 반환

```python
a = [1, 2, 3, 4, 5]
print(a.index(3))
```

* **in 키워드**
 + 리스트 내에 해당 값이 존재하는지 확인
 + value in [list]
 + True, False 중 한가지로 반환

```python
a = [1, 2, 3, 4, 5, 10]
b = 10
print(b in a)
```

* **list 정렬**
 + sort() -> 리스트 자체를 내부적으로 정렬
 + sorted() -> 리스트의 정렬된 복사본을 반환

```python
a = [9, 10, 2, 3, 1, 2, 4, 5]
b = sorted(a)
print(a)
print(b)

a.sort()
print(a)
```

### **dictionary** (가장 자주 사용)
 + 키와 값을 갖는 데이터 구조
 + 키는 내부적으로 hash값으로 저장
 + 순서를 따지지 않음. 즉, 인덱스가 없음  
 
 ※ 리스트는 원소의 개수가 아무리 많더라도 인덱스를 통해 접근하기 때문에 접근시간이 매우 빠름  
 ※ 딕셔너리는 인덱스가 없지만 hash 키를 통해서 접근할 수 있음

```python
a = {'korea': 'seoul',
    'canada': 'ottawa',
    'usa': 'washington d.c'}
print(a)
```

+ **항목 추가 및 변경**
 - 기존에 키가 존재 하면, 새로운 값으로 업데이트(동일한 키 중복으로 존재 불가-유일성)
 - 존재하지 않으면, 새로운 키, 값 생성

```python
a = {'korea': 'seoul',
    'canada': 'ottawa',
    'usa': 'washington d.c'}
a['japan'] = 'tokyo'
print(a)

a['japan'] = 'kyoto'
print(a)
```

+ **update()**
 * 두 딕셔너리를 병합함
 * 겹치는 키가 있다면 parameter로 전달되는 키 값이 overwrite된다.

```python
a = {'a': 1, 'b': 2, 'c': 3}
b = {'a': 2, 'd': 4, 'e': 5}
a.update(b)
print(a)
```

* **key삭제**
 - del 키워드 사용(범용적으로 삭제: 변수를 생성하면 노트북 파일 안에서 어디든지 사용가능한데, del로 삭제하면 접근 불가능)
 - pop 함수 이용

```python
a = {'a': 1, 'b': 2, 'c': 3}
del a['b']
print(a)

c = a.pop('a')
print(a)
print(c)
```

* **clear()**
  + 딕셔너리의 모든 값을 초기화

```python
a = {'a': 1, 'b': 2, 'c': 3}
print(a)
a.clear()
print(a)
```

* **in**
 + key값 존재 확인
 + O(1) 연산 - 딕셔너리의 크기와 관계없이 항상 연산의 속도가 일정하다는 의미

```python
a = {'a': 1, 'b': 2, 'c': 3}
print('b' in a)
print('d' in a)
```

* **value access**
 + dict[key]로 접급, 키가 없는 경우 에러 발생 
 + .get() 함수로 접근, 키가 없는 경우 None반환

```python
a = {'a': 1, 'b': 2, 'c': 3}
print(a.get('d'))
print(a.get('d', 'default value'))
# print(a['d']) # KeyError 발생
```

#### **모든 keys, values 접근**
 + keys() - 키만 반환
 + values() - 값만 반환
 + items() - 키, 값의 튜플을 반환

```python
a = {'a': 1, 'b': 2, 'c': 3}
print(a.keys())
print(a.values())
print(a.items())
```

### **tuple**
 + 리스트와 같이 복수개의 값을 갖는 컬렉션 타입
 + 생성된 후 변경이 불가능 (immutable)
 + 함수의 return값에 tuple이 자주 사용됨

```python
a = (1, 2, 3)
print(a)
# a[0] = 100 # TypeError 발생
```

### **tuple unpacking**
- 튜플의 값을 차례대로 변수에 대입
- 여러개의 변수에 동시에 값을 집어넣을 때 사용 = tuple unpacking

```python
a, b, c, d = 100, 200, 300, 400
print(a, b, c, d)
```

- 연습문제 a와 b의 값을 교환하시오.

```python
a = 4
b = 5

# 파이썬에서는 tuple unpacking을 이용하여 간단히 교환 가능
a, b = b, a 

print(a)
print(b)
```

### **set**
- dictionary에서 key만 활용하는 데이터 구조로 이해
- 수학에서의 집합과 동일한 개념
- 중복값을 허용하지 않음
- 인덱스가 없음(=순서가 없음)

```python
a = {1, 2, 3, 3, 4, 5, 5, 5}
print(a)
```

* **set()으로 집합으로 변환**

```python
a = [1, 2, 3, 4, 5, 1, 2, 3]
b = set(a)
print(b)
```

- **set operations**
  + 수학 연산과 동일
  + 교집합, 합집합, 차집합 등 지원

```python
a = {1, 2, 3}
b = {2, 3, 4}

print(a.union(b))        # 합집합
print(a.intersection(b)) # 교집합
print(a.difference(b))   # 차집합
```
