## 1. 조건문(Conditional Statement)이란?

조건문은 **특정 조건을 만족할 때만 코드를 실행**하고 싶을 때 사용합니다. '만약 ~라면, ...을 실행하라'는 논리적 흐름을 코드로 구현하는 것이죠.

모든 조건은 **참(True)** 또는 **거짓(False)**을 나타내는 `boolean` 값으로 평가됩니다. 파이썬에서는 `if`, `elif`, `else` 키워드를 사용하여 조건문을 만듭니다.

### 코드 블록과 들여쓰기

조건문 아래에 실행될 코드는 **들여쓰기(indentation)**를 통해 구분합니다. 들여쓰기 된 코드 뭉치를 **코드 블록(code block)**이라고 부릅니다. 파이썬에서는 보통 4개의 공백(space) 또는 1개의 탭(tab)을 사용해 들여쓰기를 합니다.

코드 블록의 시작을 알리는 조건문 끝에는 항상 콜론(`:`)을 붙여야 합니다.

```python
# `6 >= 5` 라는 조건이 참(True)이므로, 아래 들여쓰기 된 코드 블록이 실행됩니다.
if 6 >= 5:
    print('6은 5보다 크거나 같습니다.')
    print('이 줄도 if 조건에 포함됩니다.')

# 이 코드는 들여쓰기가 되어있지 않으므로, if 조건과 상관없이 항상 실행됩니다.
print('이 코드는 if문 바깥에 있습니다.')
```

````python
if 6 >= 5:
    print('6은 5보다 크거나 같습니다.')
    print('이 줄도 if 조건에 포함됩니다.')

print('이 코드는 if문 바깥에 있습니다.')
````

## 2. 다양한 조건문 구조

### `if-else` : 두 가지 경우만 있을 때

조건이 **참일 때**와 **거짓일 때** 각각 다른 코드를 실행하고 싶다면 `if`와 `else`를 함께 사용합니다. `else`는 '그렇지 않다면'이라는 의미를 가집니다.

**주의:** `if`와 `else` 사이에는 다른 코드가 들어갈 수 없습니다.

````python
number = 10

if number % 2 == 0:
    print('짝수입니다.')
else:
    print('홀수입니다.')
````

### `if-elif-else` : 여러 조건을 검사할 때

여러 개의 조건을 순서대로 확인하고 싶을 때는 `elif`(`else if`의 줄임말)를 사용합니다.

1. `if` 조건을 먼저 확인하고, 참이면 `if` 블록만 실행하고 전체 조건문을 빠져나갑니다.
2. `if`가 거짓이면, 첫 번째 `elif` 조건을 확인합니다. 참이면 해당 `elif` 블록만 실행하고 빠져나갑니다.
3. 앞선 모든 `if`, `elif` 조건이 거짓일 경우, 마지막 `else` 블록이 실행됩니다. (`else`는 생략 가능합니다.)

````python
score = 85

if score >= 90:
    print('A학점입니다.')
elif score >= 80:
    print('B학점입니다.')
elif score >= 70:
    print('C학점입니다.')
else:
    print('F학점입니다.')
````

## 3. 복잡한 조건 만들기: 논리 연산자

때로는 여러 조건을 조합해야 할 필요가 있습니다. 이때 **논리 연산자**(`and`, `or`, `not`)를 사용합니다.

* `and`: **모든** 조건이 참(True)이어야 전체가 참(True)이 됩니다.
* `or`: 조건 중 **하나라도** 참(True)이면 전체가 참(True)이 됩니다.
* `not`: 조건의 결과를 **반대로** 뒤집습니다. (True -> False, False -> True)

** 연산자 우선순위:** `not` > `and` > `or` (괄호를 사용하면 우선순위를 명확히 할 수 있습니다.)

````python
username = 'admin'
password = '1234'

# username이 'admin'이고(AND) password가 '1234'일 때 로그인 성공
if username == 'admin' and password == '1234':
    print('로그인에 성공했습니다.')
else:
    print('아이디 또는 비밀번호를 확인하세요.')

# 돈이 5000원 이상 있거나(OR) 쿠폰이 있을 때 택시를 탄다
money = 4000
has_coupon = True

if money >= 5000 or has_coupon:
    print('택시를 탄다.')
else:
    print('걸어간다.')
````

## 4. Boolean이 아닌 값의 진실 혹은 거짓 (Truthiness)

파이썬에서는 `True`나 `False`가 아니더라도 조건문에 사용될 수 있습니다. 이때 값 자체가 '참 같은 값(Truthy)'인지 '거짓 같은 값(Falsy)'인지 판단합니다.

####  거짓(False)으로 간주되는 값들:
* `None`
* 숫자 `0`, `0.0`
* 비어 있는 문자열: `''`, `""`
* 비어 있는 컬렉션: `[]`(리스트), `()`(튜플), `{}`(딕셔너리), `set()`(집합)

####  참(True)으로 간주되는 값들:
* **위에서 언급한 '거짓 같은 값'을 제외한 모든 값** (예: `1`, `'hello'`, `[1, 2]`, `{'a': 1}`)

````python
my_list = [1, 2, 3]

# my_list가 비어있지 않으므로 True로 평가됩니다.
if my_list:
    print('my_list에 요소가 있습니다.')
else:
    print('my_list가 비어있습니다.')

empty_str = ''

# empty_str은 비어 있으므로 False로 평가됩니다.
if not empty_str:
    print('문자열이 비어있습니다.')
````

## 5. 중첩 조건문 (Nested Condition)

하나의 조건문 안에 또 다른 조건문을 넣어 더 복잡한 논리를 구현할 수 있습니다. 바깥 조건문이 참일 경우에만 안쪽 조건문을 검사하게 됩니다.

들여쓰기를 통해 구조를 구분하며, 중첩의 깊이에는 제한이 없지만 너무 깊어지면 코드를 읽기 어려워지므로 주의해야 합니다.

````python
num = 20

if num > 0:
    print('양수입니다.')
    
    # 안쪽 조건문: 양수인 경우에만 짝수/홀수를 판별합니다.
    if num % 2 == 0:
        print('그리고 짝수입니다.')
    else:
        print('그리고 홀수입니다.')
elif num == 0:
    print('0입니다.')
else:
    print('음수입니다.')
````
