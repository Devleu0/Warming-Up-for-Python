## Ⅰ. 기본 내장 함수 (Built-in Functions)

모듈 임포트(`import`) 없이 바로 사용할 수 있는 파이썬의 기본 함수들입니다.

### `print()`
값을 화면에 출력합니다.

```python
print("Hello, Jupyter!")
```

### `input()`
사용자로부터 문자열을 입력받습니다. (노트북 환경에서는 입력창이 나타납니다)

### `len()`
객체(문자열, 리스트 등)의 길이를 반환합니다.

```python
my_list = [1, 2, 3, 4, 5]
print(f"리스트의 길이: {len(my_list)}")

my_string = "Python"
print(f"문자열의 길이: {len(my_string)}")
```

### `type()`
객체의 자료형을 확인합니다.

```python
print(type(123))          # 정수형
print(type(3.14))         # 실수형
print(type("Hello"))      # 문자열
print(type([1, 2, 3]))    # 리스트
```

### `int()`, `float()`, `str()`, `list()`
값을 다른 자료형으로 변환합니다.

```python
num_str = "100"
num_int = int(num_str)
print(f"문자열 '{num_str}' -> 정수 {num_int}")

pi_float = 3.14
pi_str = str(pi_float)
print(f"실수 {pi_float} -> 문자열 '{pi_str}'")

text = "abc"
char_list = list(text)
print(f"문자열 '{text}' -> 리스트 {char_list}")
```

### `sum()`, `max()`, `min()`
숫자로 이루어진 반복 가능한 객체(리스트 등)의 합계, 최댓값, 최솟값을 구합니다.

```python
numbers = [10, 5, 8, 20, 3]

print(f"합계: {sum(numbers)}")
print(f"최댓값: {max(numbers)}")
print(f"최솟값: {min(numbers)}")
```

### `sorted()`
객체를 정렬한 **새로운** 리스트를 반환합니다. (원본은 변경되지 않습니다)

```python
original_list = [3, 1, 4, 1, 5, 9, 2]
sorted_list = sorted(original_list)

print(f"원본 리스트: {original_list}")
print(f"정렬된 새 리스트: {sorted_list}")
```

## Ⅱ. 자료형별 내장 함수 (메서드)
각 자료형이 가지고 있는 고유한 기능의 함수(메서드)들입니다.

### 1. 문자열 (String)

```python
s = "  Hello, Python World!  "

# .upper() / .lower()
print(f"대문자: {s.upper()}")
print(f"소문자: {s.lower()}")

# .strip()
print(f"공백 제거: '{s.strip()}'")

# .replace()
print(f"문자열 교체: {s.replace('Python', 'Jupyter')}")

# .split()
words = "apple,banana,cherry"
print(f"문자열 나누기: {words.split(',')}")

# .join()
char_list = ['p', 'y', 't', 'h', 'o', 'n']
print(f"문자열 합치기: {'-'.join(char_list)}")
```

### 2. 리스트 (List)

```python
fruits = ['apple', 'banana']
print(f"초기 리스트: {fruits}")

# .append()
fruits.append('cherry')
print(f"append 후: {fruits}")

# .pop()
popped_item = fruits.pop()
print(f"pop으로 제거된 항목: {popped_item}")
print(f"pop 후: {fruits}")

# .sort() (원본 변경)
numbers = [3, 1, 4, 2]
numbers.sort()
print(f"sort 후: {numbers}")

# .reverse() (원본 변경)
numbers.reverse()
print(f"reverse 후: {numbers}")

# .insert()
numbers.insert(1, 99) # 1번 인덱스에 99 삽입
print(f"insert 후: {numbers}")
```

### 3. 딕셔너리 (Dictionary)

```python
person = {
    'name': '홍길동',
    'age': 30,
    'city': '서울'
}

# .keys()
print(f"키: {person.keys()}")

# .values()
print(f"값: {person.values()}")

# .items() (for 문과 함께 자주 사용)
print(f"항목들: {person.items()}")
for key, value in person.items():
    print(f"  - {key}: {value}")

# .get()
print(f"이름: {person.get('name')}")
print(f"직업: {person.get('job', '알 수 없음')} # 키가 없을 때 기본값 반환")
```
