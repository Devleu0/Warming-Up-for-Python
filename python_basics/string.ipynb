### **string (문자열)**
 - 복수개의 문자를 순서대로 나열 한 것
 - 문자열은 '(작은따옴표) 혹은 "(큰따옴표) 사이에 문자를 넣어서 생성
 - 문자열 자체에 ', "가 있는 경우에는 각각 그 반대의 기호로 생성

#### escape string (이스케이프 문자)
* 문자열내의 일부 문자의 의미를 달리하여 `특정한 효과`를 주는 것
* `\n` : new line `\t` : tab 등

```python
a = 'Hello\nWorld'
b = 'Hello\tWorld'
print(a)
print(b)
```

#### **indexing & slicing string (문자열 인덱스 및 추출)**
  - 문자열의 각 문자는 순서가 있음
  - 이때 각 문자열의 순서를 **인덱스** 라고 함
  - 첫번째 문자부터 마지막까지 차례대로의 순서를 가짐
  - 첫번째 시작문자의 순서는 0으로 시작 (1이 아님)

```python
a = 'Hello World'
print(a[0])
print(a[6])
```

#### **-1 인덱스**
  - 다른 언어와는 달리, python의 경우 음수 인덱스를 지원
  - -1이 가장 마지막 인덱스를, -2가 마지막에서 두번째 인덱스를 의미...

```python
a = 'Hello World'
print(a[-1])
print(a[-5])
```

#### **인덱스의 범위**
  - 인덱스는 [0, 문자열의 길이) 의 범위만 유효
  - 음수 인덱스를 사용할 경우, [-문자열의 길이, -1]
  - 범위를 넘어갈 경우 에러 발생

```python
a = 'Hello World' # 길이는 11
# print(a[11]) # 주석을 해제하면 IndexError 발생
```

#### **문자열 slicing**
  - 인덱스가 하나의 문자만을 추출한다면,
  - slicing은 부분 문자열을 추출한다고 볼 수 있음
  - [시작:끝]와 같이 명시하여 [시작, 끝-1]에 해당하는 부분 문자열을 추출
  - 시작, 끝 인덱스가 생략이 되어 있다면, 0부터 혹은 끝까지로 간주

```python
a = 'Hello World'
print(a[0:5])
print(a[:5])
print(a[6:])
print(a[:])
```

* **replace** 
 - 문자열 내의 특정 문자를 치환

```python
a = 'Hello World'
b = a.replace('World', 'Python')
print(b)
```

* **format**
 - 문자열내의 특정한 값을 변수로부터 초기화하여 동적으로 문자열을 생성 

```python
temperature = 25
s = '현재 온도는 {}도 입니다.'.format(temperature)
print(s)

# f-string (Python 3.6+)
s2 = f'현재 온도는 {temperature}도 입니다.'
print(s2)
```

* **len**
 - 문자열의 길이를 계산

```python
a = 'Hello World'
print(len(a))
```

* **split**
 - 문자열을 특정한 문자 구분하여(delimiter) 문자열의 리스트로 치환
 - ()안에 아무것도 입력하지 않으면 Default는 공백으로 split!

```python
a = 'Hello World Python'
b = a.split()
print(b)

c = 'apple,banana,orange'
d = c.split(',')
print(d)
```

* **strip**
 - 문자열 양옆의 공백제거
 - strip(): 문자열 양옆 공백제거

```python
a = '   Hello World   '
b = a.strip()
print(f'"{a}"')
print(f'"{b}"')
```

* **문자열 구성 파악하기 is...()**
 - islower(): 문자열이 소문자로 구성되어 있는지 확인
 - isupper(): 문자열이 대문자로 구성되어 있는지 확인
 - isdigit(): 문자열이 숫자로 인식될수있는지 확인
 - isalnum(): 문자열이 알파벳 또는 숫자로만 구성되어 있는지 확인
 - isdecimal(): 문자열이 정수 형태인지 확인

```python
s1 = 'hello'
s2 = 'HELLO'
s3 = '12345'
s4 = 'hello123'

print(f'"{s1}".islower(): {s1.islower()}')
print(f'"{s2}".isupper(): {s2.isupper()}')
print(f'"{s3}".isdigit(): {s3.isdigit()}')
print(f'"{s4}".isalnum(): {s4.isalnum()}')
print(f'"{s3}".isdecimal(): {s3.isdecimal()}')
```

* **find()와 rfind()**
 - 문자열 내부에 특정 문자가 어디에 위치하는지 확인(인덱스 번호 알려줌)
 - find()는 왼쪽부터, rfind()는 오른쪽부터 찾음

```python
s = 'hello world, hello python'
print(s.find('hello')) # 가장 먼저 찾은 hello의 시작 인덱스
print(s.rfind('hello')) # 가장 나중에 찾은 hello의 시작 인덱스
print(s.find('java')) # 없는 경우 -1 반환
```

* **in**
